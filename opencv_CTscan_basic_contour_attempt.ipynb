```python
ct_image = cv.imread('/content/drive/MyDrive/Colab Notebooks/freecodecamp-opencv/duke_university_CSpineSeg_screenshot 2.png')
```

```python
ct_image_displayed = cv2_imshow(ct_image)
```

```python
blur = cv.GaussianBlur(ct_image, (7,7), cv.BORDER_DEFAULT)
canny = cv.Canny(blur, 125,175)
cv2_imshow(canny)
```

```python
blur2 = cv.GaussianBlur(ct_image, (3,3), cv.BORDER_DEFAULT)
canny2 = cv.Canny(blur2, 125,175)
cv2_imshow(canny2)
#if the parenthesis value of 7,7 is changed to 2,2, we get an error. it says:
#error: OpenCV(5.0.0) /io/opencv/modules/imgproc/src/smooth.dispatch.cpp:291: error: (-215:Assertion failed)
# ksize.width > 0 && ksize.width % 2 == 1 && ksize.height > 0 && ksize.height % 2 == 1 in function 'createGaussianKernels'
#--------
#whatever value of width we choose needs to have a remainder of 1 when divided by 2 which is what the modulo ksize.width %2 ==1 says.
#the ksize.width and ksize.height can only be odd numbers? Why?
```

```python
blur3 = cv.GaussianBlur(ct_image, (1,1), cv.BORDER_DEFAULT)
canny3 = cv.Canny(blur3, 125,175)
cv2_imshow(canny3)
```

```python
gray = cv.cvtColor(ct_image, cv.COLOR_BGR2GRAY)
threshold, thresh = cv.threshold(gray, 50, 255, cv.THRESH_BINARY)
cv2_imshow(thresh)
```

```python

threshold2, thresh2 = cv.threshold(gray, 100, 255, cv.THRESH_BINARY)
cv2_imshow(thresh2)
```

```python
adaptive_thresh =cv.adaptiveThreshold(gray, 255,cv.ADAPTIVE_THRESH_MEAN_C, cv.THRESH_BINARY, 11, 9 )
cv2_imshow(adaptive_thresh)
```
